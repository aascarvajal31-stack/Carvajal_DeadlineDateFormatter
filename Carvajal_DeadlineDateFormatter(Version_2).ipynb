```python
import datetime

event_title = input("Enter event title: ")
year = int(input("Enter year: "))
month = int(input("Enter month (numerical): "))
day = int(input("Enter day: "))

event_info = [event_title, year, month, day]
event_date = datetime.date(event_info[1], event_info[2], event_info[3])

month_name = event_date.strftime("%B")
weekday = event_date.strftime("%A")

def event_info_display():
  for i in range(len(event_info)):
    if i == 0:
      print("Event: ", event_info[i].upper())
    if i == 1:
      print("Date: ", month_name, event_info[3], ",", event_info[1])
      print("Year: ", event_info[i])
    if i == 2:
      print("Month: ", month_name)
    if i == 3:
      print("Day: ", event_info[i])
      print("Weekday: ", weekday)

print("\n--Formatted Output-- ")
print("\nEvent Information")
event_info_display()
```
